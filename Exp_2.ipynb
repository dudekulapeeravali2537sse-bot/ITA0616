```python

import csv

# Read training data from CSV file
with open("training_data.csv", "r") as file:
    data = list(csv.reader(file))

# Separate attributes and examples
attributes = data[0][:-1]
examples = data[1:]

n = len(attributes)

# Initialize S and G boundaries
S = ['0'] * n
G = [['?'] * n]

# Check whether hypothesis covers an example
def covers(h, x):
    return all(h[i] == '?' or h[i] == x[i]
               for i in range(n))

# Check whether h1 is more general than h2
def more_general(h1, h2):
    return all(a == '?' or a == b
               for a, b in zip(h1, h2))

# Candidate Elimination Algorithm
for example in examples:

    x = example[:-1]
    target = example[-1]

    if target == 'Yes':

        # Remove inconsistent hypotheses from G
        G = [g for g in G if covers(g, x)]

        # Generalize S minimally
        if S == ['0'] * n:
            S = x.copy()
        else:
            for i in range(n):
                if S[i] != x[i]:
                    S[i] = '?'

        # Remove redundant hypotheses from G
        G = [g for g in G if more_general(g, S)]

    elif target == 'No':

        new_G = []

        # Specialize inconsistent hypotheses in G
        for g in G:

            if covers(g, x):

                for i in range(n):

                    if g[i] == '?':

                        values = set(row[i] for row in examples)

                        for value in values:

                            if value != x[i] and (
                                S[i] == '?' or S[i] == value
                            ):
                                h = g.copy()
                                h[i] = value

                                if more_general(h, S):
                                    new_G.append(h)

            else:
                new_G.append(g)

        G = new_G

    print("\nTraining Example:", example)
    print("Specific Boundary S:", S)
    print("General Boundary G:", G)

# Display final boundaries
print("\n========== FINAL RESULT ==========")

print("\nSpecific Boundary (S):")
for h in [S]:
    print(h)

print("\nGeneral Boundary (G):")
for h in G:
    print(h)

```
