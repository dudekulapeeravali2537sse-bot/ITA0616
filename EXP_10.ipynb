```python

# Program to implement Expectation Maximization Algorithm

import numpy as np
import matplotlib.pyplot as plt
from sklearn.mixture import GaussianMixture

# Step 1: Create sample dataset
X = np.array([
    [1.0], [1.5], [2.0], [2.5], [3.0],
    [8.0], [8.5], [9.0], [9.5], [10.0]
])

# Step 2: Create Gaussian Mixture Model
gmm = GaussianMixture(
    n_components=2,
    random_state=42
)

# Step 3: Fit the model using EM algorithm
gmm.fit(X)

# Step 4: Predict cluster labels
labels = gmm.predict(X)

# Step 5: Display results
print("Cluster Labels:")
print(labels)

print("\nMean of each Gaussian:")
print(gmm.means_)

print("\nCovariance of each Gaussian:")
print(gmm.covariances_)

print("\nMixing Weights:")
print(gmm.weights_)

# Step 6: Plot the clusters
plt.scatter(X[:, 0], np.zeros(len(X)),
            c=labels, cmap='viridis', s=100)

plt.xlabel("Data Values")
plt.ylabel("Clusters")
plt.title("Expectation Maximization Algorithm")
plt.show()
```
